```python
import pandas as pd
import psycopg2
from sqlalchemy import create_engine
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import seaborn as sns
```

```python
# DB接続情報設定
engin = create_engine("postgresql+psycopg2://user:password@job_data_db:5432/job_data")
```

```python
# 各TBLデータ抽出クエリ設定
query_job_postings = "select * from job_data_analysis.tb_job_postings"
query_job_postings_skills = "select * from job_data_analysis.tb_job_postings_skills"
query_skills = "select * from job_data_analysis.tb_skills"
```

```python
# 各TBLデータ格納
df_job_postings = pd.read_sql(query_job_postings, engin)
df_job_postings_skills = pd.read_sql(query_job_postings_skills, engin)
df_skills = pd.read_sql(query_skills, engin)
```

### 1.要約統計量

```python
# 給与データの可視化
plt.figure(figsize=(10, 6))
sns.set(style="whitegrid")

plt.figure(figsize=(10, 6))
sns.boxplot(df_job_postings['salary'])

# タイトル・ラベル
plt.title('salary_boxplot')
plt.xlabel('annual income')

plt.show()
```
